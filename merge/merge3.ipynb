```python
import pandas as pd
import numpy as np

df_transacciones=pd.read_csv("../data/clean/features/hey_transacciones_features_limpio.csv")

df_transacciones.head()
```

```python
df_transacciones.info()
```

```python
df_clientes=pd.read_csv("../data/clean/features/hey_clientes_features_limpio.csv")

df_clientes.head()
```

```python
df_clientes.info()
```

```python
df_productos=pd.read_csv("../data/clean/features/hey_productos_features_limpio.csv")

df_productos.head()
```

```python
df_productos.info()
```

```python
# Verificamos que no haya duplicados en la llave primaria
assert df_clientes['user_id'].is_unique, "¡Alerta! Hay user_id duplicados en Clientes"
assert df_transacciones['user_id'].is_unique, "¡Alerta! Hay user_id duplicados en Transacciones"
assert df_productos['user_id'].is_unique, "¡Alerta! Hay user_id duplicados en Productos"
```

```python
# 1. Unir Clientes con Productos
df_intermedio = pd.merge(df_clientes, df_productos, on='user_id', how='inner')

# 2. Unir el resultado con Transacciones
df_master = pd.merge(df_intermedio, df_transacciones, on='user_id', how='inner')

# 3. Comprobación final
print(f"Dimensiones de la matriz final: {df_master.shape}")
```

```python
ruta_salida = "../data/clean/features/hey_merge_features_limpio.csv"

df_master.to_csv(ruta_salida, index=False, encoding="utf-8-sig")

print(f"✅ Proceso completado. Archivo guardado en: {ruta_salida}")
```
